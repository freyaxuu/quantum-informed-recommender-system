```python
from qiskit import QuantumCircuit
import numpy as np

n = 5
L = 3
qc = QuantumCircuit(n)

for l in range(L):
    for j in range(n):
        qc.rx(np.random.rand(), j)
        qc.rz(np.random.rand(), j)
        qc.ry(np.random.rand(), j)
    
    for j in range(n - 1):
        qc.cz(j, j + 1)

qc.draw('mpl')
```
